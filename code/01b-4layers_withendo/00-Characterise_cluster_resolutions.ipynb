One of the goals of this analysis is to select the resolutions that are most compatible with the layer partitioning performed in MRI. The MRI partitioning is based on splitting equally the number of voxels across each layer, and not based on features of the cortext. Therefore, the chosen partitioning on this level must attempt to approximate this. 

**2 cortical layers**
 - The best resolution for 2 cortical layers is 'bayesSpace_harmony_4', where clusters 1 and 2 would be excluded. 

**3 cortical layers**
 - The best resolution for 3 cortical layers is 'bayesSpace_harmony_6', where clusters 1, 4 and 6 would be excluded. The ratio between the largest cortical cluster for harmony_6 is 1.84, while for harmony_5 is 1.98. 

**4 cortical layers**
 - The best resolution for 4 cortical layers is 'bayesSpace_harmony_7', where clusters 1, 5 and 7 would be excluded. The ratio between the largest cortical layer and the smallest is 1.33
