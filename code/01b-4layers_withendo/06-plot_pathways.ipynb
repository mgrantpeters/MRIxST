```python
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob, os
```

```python
paths = glob.glob('../../processed_data/01b-4layers_withendo/05-pathway_scoring/scores_4*.csv')
for pathway in paths:
    df = pd.read_csv(pathway, index_col=0)
    sns.boxplot(data=df, x="bayesSpace_harmony_7", y="score", palette = 'Set2', order = ['WM', 'L1','L2','L3', 'L4'])
    sns.swarmplot(data=df, x="bayesSpace_harmony_7", y="score", color = 'black', order = ['WM', 'L1','L2','L3', 'L4'])
    plt.title(pathway.split('.csv')[0].split('GO_terms_')[1])
    #plt.savefig('../../plots/01b-4layers_withendo/06-plot_pathways/_%s.png' % cell)
    plt.show()
```
