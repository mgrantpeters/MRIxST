```python
###--------------------------------------------LOAD LIBRARIES
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import copy
```

```python
nonIF =  sc.read("../../LIBD_LR/grantpm/DLPFC_Visium_LIBD/processed-data/MGP_analysis/nonIF_c2l_anndata_combined_30samples.h5ad")
```

```python
# Prepare data

subset = nonIF[nonIF.obs['bayesSpace_harmony_7'].isin([2,3,4,5,6])]   # Excluding WM and meningeal layers

subset.obs['bayesSpace_harmony_7'] = subset.obs['bayesSpace_harmony_7'].astype('category').cat.rename_categories({   # Renaming to match order of MRI (L1, closer to WM, L4 closer to meninges)
                                                     2 : 'L4', 
                                                     3 : 'L3', 
                                                     4 : 'L2', 
                                                     5 : 'WM',
                                                     6 : 'L1'})

subset.obs['sample_id'] = subset.obs['sample_id'].astype(str)

df_deconvo = subset.obs[np.isnan(subset.obs['Oligo'])==False]

deconvo = ['Astro',
 'EndoMural',
 'Excit_L2_3',
 'Excit_L3',
 'Excit_L3_4_5',
 'Excit_L4',
 'Excit_L5',
 'Excit_L5_6',
 'Excit_L6',
 'Inhib',
 'Micro',
 'OPC',
 'Oligo']

#summarised = subset.obs.groupby(['sample_id', 'bayesSpace_harmony_7']).median()[['Astro', 'EndoMural', 'Excit_L2_3',
#       'Excit_L3', 'Excit_L3_4_5', 'Excit_L4', 'Excit_L5', 'Excit_L5_6',
#       'Excit_L6', 'Inhib', 'Micro', 'OPC', 'Oligo']].reset_index()

probs = subset.obs[['Astro', 'EndoMural', 'Micro', 'Oligo', 'OPC', 'Excit_L2_3', 'Excit_L3',
       'Excit_L3_4_5', 'Excit_L4', 'Excit_L5', 'Excit_L5_6', 'Excit_L6',
       'Inhib']]
hot1 = copy.copy(probs)
hot1.iloc[:,:] = 0
for m in range (0,np.shape(probs)[0]):
    hot1.iloc[m,:][probs.T[probs.index[m]].nlargest(3).index]=1

annotated = hot1.merge(subset.obs[['sample_id','bayesSpace_harmony_7']], left_index = True, right_index = True, how = 'inner')
annotated = annotated.groupby(['sample_id', 'bayesSpace_harmony_7']).sum().reset_index().melt(id_vars = ['sample_id', 'bayesSpace_harmony_7'])
annotated.to_csv('../results/celltype_counts/top4_cells_per_layer_count.csv')
```

```python
sns.boxplot(data = annotated[annotated['variable']=='Astro'], x = 'bayesSpace_harmony_7', y = 'value')
sns.swarmplot(data = annotated[annotated['variable']=='Astro'], x = 'bayesSpace_harmony_7', y = 'value', color = 'black')
plt.show()
```

```python
sns.boxplot(data = annotated[annotated['variable']=='Micro'], x = 'bayesSpace_harmony_7', y = 'value')
sns.swarmplot(data = annotated[annotated['variable']=='Micro'], x = 'bayesSpace_harmony_7', y = 'value', color = 'black')
plt.show()
```
