```python
import pandas as pd
import seaborn as sns
```

```python
df_PD = pd.read_csv('~/MinaRyten/Melissa/MRIxST/plots/04b-4layers_PDpredictions_withendo/Projected_PD_cell_counts.csv')
df_PD['Group'] = 'PD'
```

```python
df_ctrl = pd.read_csv('~/MinaRyten/Melissa/MRIxST/processed_data/01b-4layers_withendo/02-top3_cells_counts/count_Oligo_4SpD_withEndo.csv')
df_ctrl['region'] = df_ctrl['sample_id'].str.split('_', expand = True)[1]
grouped = df_ctrl.groupby('region').median()
grouped = grouped.reset_index().melt(id_vars=['region'])
grouped.replace({'ant':'anterior', 'mid':'middle', 'post':'posterior', 'WM':'SWM'}, inplace = True)
grouped.rename(columns={'variable':'SpD', 'value':'Oligo_PD_predicted'}, inplace = True)
grouped['Group'] = 'Control'
```

```python
merge = pd.concat([df_PD, grouped])[['SpD','region','Oligo_PD_predicted', 'Group']]
merge = merge[merge['region']=='anterior']
sns.barplot(x = merge['SpD'], y = merge['Oligo_PD_predicted'], hue = merge['Group'])
```

```python
merge = pd.concat([df_PD, grouped])[['SpD','region','Oligo_PD_predicted', 'Group']]
merge = merge[merge['region']=='middle']
sns.barplot(x = merge['SpD'], y = merge['Oligo_PD_predicted'], hue = merge['Group'])
```

```python
merge = pd.concat([df_PD, grouped])[['SpD','region','Oligo_PD_predicted', 'Group']]
merge = merge[merge['region']=='posterior']
sns.barplot(x = merge['SpD'], y = merge['Oligo_PD_predicted'], hue = merge['Group'])
```
