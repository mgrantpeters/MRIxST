```python
df = pd.read_csv("../../processed_data/07_genetic_enrichment/PD_gene_enrichment.csv")
```

```python
import requests
def fetch_go_term_name(go_code):
    url = f"https://www.ebi.ac.uk/QuickGO/services/ontology/go/terms/{go_code}"
    response = requests.get(url, headers={"Accept": "application/json"})
    
    if response.status_code == 200:
        data = response.json()
        term_name = data.get("results")[0].get("name")
        return term_name
    else:
        print("Error fetching the GO term")
        return None
```

```python
pathway_name = []
for row in df['pathway']:
    term_name = fetch_go_term_name(row)
    pathway_name.append(term_name)
df['Pathway name'] = pathway_name
```
