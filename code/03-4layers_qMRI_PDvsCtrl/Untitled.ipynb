```python
import pandas as pd
from statsmodels.stats.multitest import fdrcorrection as fdr
```

```python
df = pd.read_csv('../../stats/03-4layers_PDvsCtrl/stats_R1_group_comparison_PDvsCtrl.csv').loc[10:14]

df['padj'] = list(pd.DataFrame(fdr(df['p.value'], alpha=0.05, method='indep', is_sorted=False)).T[1])
```
