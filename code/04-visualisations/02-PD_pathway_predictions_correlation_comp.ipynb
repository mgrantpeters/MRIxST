```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
import glob
import os
from statsmodels.stats.multitest import fdrcorrection as fdr
```

```python
custom_palette = {
    'SpD1': '#ED5E70',  # blue
    'SpD2': '#CCBB44',  # orange
    'SpD3': '#238934',  # green
    'SpD4': '#66CCEE'   # red
}

custom_palette_CT = {
    'SpD1': '#e8bec4',  # blue
    'SpD2': '#e8e1ae',  # orange
    'SpD3': '#b2bfb4',  # green
    'SpD4': '#c2e2ed'   # red
}

marker_dict = {
    'anterior': 'o',   # circle
    'middle': '^',     # triangle
    'posterior': 's'   # square
}

sns.set_style("whitegrid")
```

```python
df_PD = pd.read_csv('/home/MRMelissaGrantPeters/MinaRyten/Melissa/MRIxST/stats/04c-4layers_PDpredictions_withendo_with_PrD/Projected_PD_pathway_scores_no_SWM.csv')
df_PD['SpDregion'] = df_PD['SpD']+df_PD['region']
df_PD = df_PD.set_index('SpDregion')

df_CT = pd.read_csv('/home/MRMelissaGrantPeters/MinaRyten/Melissa/MRIxST/stats/02c-4layers_qMRIvsST_withendo_with_PrD/LOO_predictions_control_qMRI_vs_GO_POI_no_SWM.csv')
df_CT['SpDregion'] = df_CT['SpD']+df_CT['region']
df_CT = df_CT.set_index('SpDregion')

paths = glob.glob('../../processed_data/01b-4layers_withendo/05-pathway_scoring/scores_4*.csv')

interest_paths = ['GO0006954','GO0048713','GO0006631', 'GO0006682', 'GO0014808','GO0030173', 'GO0033280', 'GO0034625','GO0036092', 'GO0042382', 'GO0051861','GO0071007', 'GO0090556', 'GO1990531']
```

```python
for pathway_code in interest_paths:
    
    predicted = df_PD.loc[:, df_PD.columns.str.contains(pathway_code)]
    col_name = str(df_PD.loc[:, df_PD.columns.str.contains(pathway_code)].columns[0])
    predicted.rename(columns = {col_name : 'PD_predicted'}, inplace = True)
    real = pd.read_csv([item for item in paths if pathway_code in item][0], index_col = 0)

    # Create region column
    conditions = [
        real['sample_id'].str.contains('ant'),
        real['sample_id'].str.contains('mid'),
        real['sample_id'].str.contains('post')
    ]
    choices = ['anterior', 'middle', 'posterior']

    real['region'] = np.select(conditions, choices, default='unknown')

    # Rename SpD
    real.rename(columns={'bayesSpace_harmony_7':'SpD'}, inplace = True)

    # Combine SpD and region
    real['SpDregion'] = real['SpD']+real['region']
    med_real = real.groupby('SpDregion').median()
    
    #Join PD predicted and control real
    merged = predicted.merge(med_real, left_index = True, right_index = True, how = 'inner')
    res = []
    
    merged['SpD'] = merged.index.str[:4]
    merged['region'] = merged.index.str[4:]
    
    print(pathway_code)
    
    # Subset data
    
    X = np.asarray(merged["score"]).reshape(-1, 1)
    Y = np.asarray(merged["PD_predicted"]).reshape(-1, 1)
    
    # Fit model
    model = LinearRegression().fit(X, Y)
    
    # R² score
    r_squared = model.score(X, Y)
    print("R²:", r_squared)
    
    
    # Add intercept
    X_sm = sm.add_constant(X)
    sm_model = sm.OLS(Y, X_sm).fit()
    p_value = sm_model.pvalues[1]  # p-value for the slope
    print("p-value:", p_value)
    
    r2_text = f"$R^2$ = {r_squared:.3f}"
    pval_text = f"p = {p_value:.3e}"

    res.append(pathway_code)
    res.append(r_squared)
    res.append(p_value)
    
    # Plot
    plt.figure(figsize=(5, 4))
    plt.plot(X, model.predict(X), color='darkgrey', label='Linear fit')
    ax = sns.scatterplot(x=merged["score"], y=merged["PD_predicted"], hue=merged["SpD"], palette = custom_palette, style=merged["region"], markers=marker_dict)
    ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
    plt.title(f"{pathway_code}")
    plt.ylabel("PD qMRI-predicted GO pathway score")
    plt.xlabel("SRT GO pathway score")
    plt.tight_layout()
    
    plt.text(
    x=0.05, y=0.95,               # relative position in axes coordinates
    s=f"{r2_text}\n{pval_text}", # multiline annotation
    transform=plt.gca().transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="gray"))
    plt.savefig('../../plots/04-visualisations/%s_PD_prediction.png' % pathway_code, dpi = 300, bbox_inches = 'tight')
    plt.savefig('../../plots/04-visualisations/%s_PD_prediction.pdf' % pathway_code, bbox_inches = 'tight')
    plt.show()
    
    if (pathway_code == interest_paths[0]):
        full_results = pd.DataFrame(res).T
    else:
        full_results = pd.concat([full_results, pd.DataFrame(res).T])
```

```python
full_results.rename(columns={0:'cell_type',1:'R2',2:'p-value'}, inplace = True)
full_results['padj'] = fdr(full_results['p-value'], alpha=0.05, method='indep', is_sorted=False)[1]
full_results.to_csv('../../processed_data/04-visualisations/PDprediction_SRT_GOBP_correlation.csv')
```

```python
pvals_beta = []
pvals_intercept = []
z_beta = []
z_int = []
for pathway_code in interest_paths:
    
    predicted_PD = df_PD.loc[:, df_PD.columns.str.contains(pathway_code)]
    col_name_PD = str(df_PD.loc[:, df_PD.columns.str.contains(pathway_code)].columns[0])
    predicted_PD.rename(columns = {col_name_PD : 'PD_predicted'}, inplace = True)
    real_PD = pd.read_csv([item for item in paths if pathway_code in item][0], index_col = 0)

    # Create region column
    conditions = [real_PD['sample_id'].str.contains('ant'),
        real_PD['sample_id'].str.contains('mid'),
        real_PD['sample_id'].str.contains('post')]
    choices = ['anterior', 'middle', 'posterior']

    real_PD['region'] = np.select(conditions, choices, default='unknown')

    # Rename SpD
    real_PD.rename(columns={'bayesSpace_harmony_7':'SpD'}, inplace = True)

    # Combine SpD and region
    real_PD['SpDregion'] = real_PD['SpD']+real_PD['region']
    med_real_PD = real_PD.groupby('SpDregion').median()
    
    #Join PD predicted and control real
    merged_PD = predicted_PD.merge(med_real_PD, left_index = True, right_index = True, how = 'inner')
    res_PD = []
    
    merged_PD['SpD'] = merged_PD.index.str[:4]
    merged_PD['region'] = merged_PD.index.str[4:]
    
    print(pathway_code)
    
    # Subset data
    Y_PD = np.asarray(merged_PD["PD_predicted"]).reshape(-1, 1)
    X_PD = np.asarray(merged_PD["score"]).reshape(-1, 1)
    
    # Fit model
    model_PD = LinearRegression().fit(X_PD, Y_PD)
    
    ### Do the same for Controls
    predicted_CT = df_CT.loc[:, df_CT.columns.str.contains(pathway_code)]
    col_name_CT = str(df_CT.loc[:, df_CT.columns.str.contains(pathway_code)].columns[0])
    predicted_CT.rename(columns = {col_name_CT : 'CT_predicted'}, inplace = True)
    real_CT = pd.read_csv([item for item in paths if pathway_code in item][0], index_col = 0)

    # Create region column
    conditions = [real_CT['sample_id'].str.contains('ant'),
        real_CT['sample_id'].str.contains('mid'),
        real_CT['sample_id'].str.contains('post')]
    choices = ['anterior', 'middle', 'posterior']

    real_CT['region'] = np.select(conditions, choices, default='unknown')

    # Rename SpD
    real_CT.rename(columns={'bayesSpace_harmony_7':'SpD'}, inplace = True)

    # Combine SpD and region
    real_CT['SpDregion'] = real_CT['SpD']+real_CT['region']
    med_real_CT = real_CT.groupby('SpDregion').median()
    
    #Join PD predicted and control real
    merged_CT = predicted_CT.merge(med_real_CT, left_index = True, right_index = True, how = 'inner')
    res_CT = []
    
    merged_CT['SpD'] = merged_CT.index.str[:4]
    merged_CT['region'] = merged_CT.index.str[4:]
    
    print(pathway_code)
    
    # Subset data
    Y_CT = np.asarray(merged_CT["CT_predicted"]).reshape(-1, 1)
    X_CT = np.asarray(merged_CT["score"]).reshape(-1, 1)
    
    # Fit model
    model_CT = LinearRegression().fit(X_CT, Y_CT)
    
    
    # Plot
    plt.figure(figsize=(4.5, 4))
    plt.plot(X_PD, model_PD.predict(X_PD), color='#4B8D86', label='Linear fit')
    ax1 = sns.scatterplot(y=merged_PD["PD_predicted"], x=merged_PD["score"], hue=merged_PD["SpD"], palette = custom_palette, style=merged_PD["region"], markers=marker_dict, edgecolor='black')
    plt.plot(X_CT, model_CT.predict(X_CT), color='#C3B3A1', label='Linear fit')
    ax2 = sns.scatterplot(y=merged_CT["CT_predicted"], x=merged_CT["score"], hue=merged_CT["SpD"], palette = custom_palette_CT, style=merged_CT["region"], markers=marker_dict, alpha = 0.7)

    ax1.legend(loc='upper left', bbox_to_anchor=(1, 1))
    plt.title(f"{pathway_code}")
    plt.ylabel("qMRI predicted score")
    plt.xlabel("Control SRT score")
    plt.tight_layout()
    plt.savefig('../../plots/04-visualisations/%s_PDpred_v_CT_SRT.png' % pathway_code, dpi = 300, bbox_inches = 'tight')
    plt.savefig('../../plots/04-visualisations/%s_PDpred_v_CT_SRT.pdf' % pathway_code, bbox_inches = 'tight')
    plt.show()
    
    beta_PD = model_PD.coef_[0][0] # Get coeficient
    intercept_PD = model_PD.intercept_[0]

    beta_CT = model_CT.coef_[0][0]
    intercept_CT = model_CT.intercept_[0]

    print(f"{pathway_code} | PD beta: {beta_PD:.3f}, CT beta: {beta_CT:.3f}")
    
    # Intercept Z test score 
    import statsmodels.api as sm
    X_PD_sm = sm.add_constant(X_PD)
    sm_model_PD = sm.OLS(Y_PD, X_PD_sm).fit()

    X_CT_sm = sm.add_constant(X_CT)
    sm_model_CT = sm.OLS(Y_CT, X_CT_sm).fit()

    se_PD = sm_model_PD.bse[1]  # slope SE
    se_CT = sm_model_CT.bse[1]

    z = (beta_PD - beta_CT) / np.sqrt(se_PD**2 + se_CT**2)
    
    #p_intercept = 2 * (1 - stats.norm.cdf(abs(z)))
    
    print(f"Z-test statistic: {z:.3f}")
    
    
    # Extract intercepts and their standard errors
    intercept_PD = sm_model_PD.params[0]      # Intercept for PD
    intercept_CT = sm_model_CT.params[0]      # Intercept for CT
    se_intercept_PD = sm_model_PD.bse[0]      # SE for PD intercept
    se_intercept_CT = sm_model_CT.bse[0]      # SE for CT intercept

    # Compute Z-test for intercept difference
    z_intercept = (intercept_PD - intercept_CT) / np.sqrt(se_intercept_PD**2 + se_intercept_CT**2)

    # Optionally compute two-tailed p-value
    #p_intercept = 2 * (1 - stats.norm.cdf(abs(z_intercept)))

    print(f"Intercept Z-test statistic: {z_intercept:.3f}")

    
    df_all = pd.concat([
    merged_PD.assign(group="PD"),
    merged_CT.assign(group="CT")
    ])

    X = sm.add_constant(df_all[["score", "group"]])
    # Encode group as dummy variable
    df_all["group_dummy"] = (df_all["group"] == "PD").astype(int)
    X = sm.add_constant(pd.DataFrame({
        "Control_real": df_all["score"],
        "Group": df_all["group_dummy"],
        "Interaction": df_all["score"] * df_all["group_dummy"]
    }))
    df_all["Predicted counts"] = df_all["PD_predicted"].fillna(df_all["CT_predicted"])
    y = df_all["Predicted counts"]

    sm_model = sm.OLS(y, X).fit()
    print(sm_model.summary())
    pvals_intercept.append(sm_model.pvalues[2])
    pvals_beta.append(sm_model.pvalues[3])
    z_beta.append(z)
    z_int.append(z_intercept)
    
```

```python
# Paraspeckles and Phosphatidylinositol have different slopes 

from statsmodels.stats.multitest import fdrcorrection as fdr

results_fit = pd.DataFrame(pvals_beta, index = interest_paths).rename(columns = {0: 'pvals_beta'})
results_fit['padj_beta'] = fdr(results_fit['pvals_beta'], alpha=0.05, method='indep', is_sorted=False)[1]
results_fit['z_beta'] = z_beta
results_fit['pvals_intercept'] = pvals_intercept
results_fit['padj_intercept'] = fdr(results_fit['pvals_intercept'], alpha=0.05, method='indep', is_sorted=False)[1]
results_fit['z_intercept'] = z_int
results_fit.to_csv('../../processed_data/04-visualisations/control_PD_GOBP_fit_comparison.csv')
```
