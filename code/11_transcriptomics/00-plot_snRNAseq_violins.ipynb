```python
meta = pd.read_csv('~/MinaRyten/Melissa/teamWood/Abundance_final_data/processed_data/covariates/covariates_All_Abundance_raw_all_cortex_annotation_level_1_All.csv', index_col = 0)
meta
```

```python
merged = df.merge(meta, left_on = 'sample_id', right_index = True, how = 'inner')
```

```python
all_clust = merged[merged['annotation_level_1']=='OG']

#print(len(all_clust['sample_id'].unique()))
#all_clust = all_clust[all_clust['% of nuclei']<all_clust['% of nuclei'].std()*4]
#print(len(all_clust['sample_id'].unique()))

color_box = 'white'

# Define sex-based colors
sex_colors = {'M': '#4B8D86', 'F': 'deeppink'}

# Create base violinplot
ax = sns.violinplot(
    data=all_clust,
    x='annotation_level_1',
    y="% of nuclei",
    hue="Group",
    palette={"Control": color_box, "PD": color_box},
    hue_order=['Control', 'PD'],
    inner='quart',
    dodge=True
)

# Get positions for dodge
group_order = ['Control', 'PD']
x_order = all_clust['annotation_level_1'].unique()

# Loop through each category and plot manually
for i, x_cat in enumerate(x_order):
    for j, group in enumerate(group_order):
        subset = all_clust[
            (all_clust['annotation_level_1'] == x_cat) &
            (all_clust['Group'] == group)
        ]
        # Calculate dodge position
        x_pos = i - 0.2 + j * 0.4  # adjust dodge width as needed

        # Plot each point with sex-based color
        for _, row in subset.iterrows():
            ax.scatter(
                x_pos,
                row["% of nuclei"],
                color=sex_colors[row['sex']],
                alpha=0.65,
                marker='D',
                s=30,
                edgecolor='black',
                linewidth=0.3
            )

plt.legend(bbox_to_anchor=(1, 1), loc='upper left', borderaxespad=0.1)
#plt.xticks(rotation=0)
plt.xlabel('')
sns.despine(right=True)
#plt.savefig('../../plots/08_validation_oligos/all_OGs_violin.pdf', dpi = 300, bbox_inches = 'tight')
#plt.savefig('../../plots/08_validation_oligos/all_OGs_violin.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```
