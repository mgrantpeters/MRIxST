```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import rcParams
```

```python
df = pd.read_csv("~/MinaRyten/Melissa/teamWood/ClusterProfiler/07-clusterProfiler_res_logfc1.csv")
df = df[df['p.adjust']<0.05]
df = df[df['annot_level'] == 2]


# figure size in inches
rcParams['figure.figsize'] = 3,5
sns.set(font_scale = 2)
summmary = df.groupby(['assay']).sum()
sns.heatmap(summmary[['Count']], cmap = 'Reds', annot=True, fmt='.4g')
plt.xlabel('')
plt.ylabel('')
plt.yticks(rotation=0)
plt.savefig('../../plots/08_validation_oligos/pathway_counts_per_celltype.png', dpi = 300, bbox_inches = 'tight')
plt.savefig('../../plots/08_validation_oligos/pathway_counts_per_celltype.pdf', dpi = 300, bbox_inches = 'tight')
plt.show()
```
